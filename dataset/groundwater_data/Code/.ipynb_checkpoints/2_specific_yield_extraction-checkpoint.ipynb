<h1>Extract aquifer type data</h1>

```python
bhanja_aquifertype.rio.crs
```

```python
lat_list = CGWB.Latitude
```

```python
long_list = CGWB.Longitude
```

```python
lat_list
```

```python
long_list
```

```python
i = 0
val_list = list()

for i in tqdm(range(len(lat_list))):
    lat = lat_list[i]
    long = long_list[i]
    
    try:
        val = bhanja_aquifertype.sel(y = lat, x = long, method = 'nearest').values
    except:
        val = np.nan
    
    val_list.append(val)
```

```python
val_list_np = np.array(val_list)
```

```python
val_list_np.shape
```

```python
len(val_list_np)
```

```python
import sklearn.cluster as cluster

# kmeans = cluster.KMeans(n_clusters=6, init='k-means++')

kmeans = cluster.KMeans(n_clusters=8, init='k-means++',random_state=42)

kmeans = kmeans.fit(val_list_np)
```

```python
kmeans.cluster_centers_
```

```python
len(kmeans.labels_)
```

```python
set(kmeans.labels_)
```

```python
kmeans_label = kmeans.labels_
```

```python
set(kmeans_label)
```

<h2>Ignore dark and black </h2>

```python
# Reclassify and compute mean specific yield
reclassify_kmeans_label = np.array(kmeans_label, dtype='f')

reclassify_kmeans_label[kmeans_label == 0] = 0.02   #Joined crystalline (cyan)  
reclassify_kmeans_label[kmeans_label == 1]= 0.13    #Unconsolidated Sedimentary  (Yellow)  
reclassify_kmeans_label[kmeans_label == 2]= 0.023   #Fractured Crystalline (Gray) 
reclassify_kmeans_label[kmeans_label == 3] = np.nan #To be ignored (black)
reclassify_kmeans_label[kmeans_label == 4] = 0.043  #Consolidated Sedimentary, permeable (Light blue)
reclassify_kmeans_label[kmeans_label == 5]= np.nan  #Black gray ignore    
reclassify_kmeans_label[kmeans_label == 6]= 0.018   #Folded metasediments (Pink) 
reclassify_kmeans_label[kmeans_label == 7] = 0.018  #Sedimentary aquitards (Orange)


#Cyan, Grey, Yellow
# Joint crystalline aquifer, Fractured crystalline, unconsolidated sedimentary aquifer
```

```python
reclassify_kmeans_label
```

```python
np.sum(np.isnan(reclassify_kmeans_label))
```

```python
np.count_nonzero(reclassify_kmeans_label)
```
